```python
import numpy as np

import pandas as pd
voltage = np.linspace(3, 4.2, 1000)


df = pd.DataFrame(voltage, columns = ['voltage'])
```

```python
from pathlib import Path

import os
import random

folder = Path(r"C:\Users\sures\Documents\DA Projects\IowaStateCells\Q_interpolated\Release")
fraction = 0.6

files = os.listdir(folder)
random.shuffle(files)

split = int(len(files) * fraction)

train_files = files[:split]
test_files = files[split:]
```

```python
print(len(train_files))

train_files[0]
```

```python
new_columns = {}

for file in train_files:
    file_path = folder / file
    try:

        df_rpt = pd.read_csv(file_path, header = None)

        Q_0 = df_rpt[0]

        Q_4 = df_rpt[3] # Not exactly right. Since I have not accomodated the fact that group 20 and beyond have additional RPT at 0.5 Weeks.

        delta_dQ = np.gradient((Q_4 - Q_0), voltage)
        new_columns[file] = delta_dQ
        
    except KeyError:
        print(f'{file} does not have week four RPT')

df = pd.concat([df, pd.DataFrame(new_columns)], axis=1)

    
    
```

```python
df.head()
```

```python
import get_lifetime
start_voltage = 3.6
end_voltage = 3.9

step_size = 0.02

mask = (voltage > start_voltage) & (voltage < end_voltage)

df_correlation = pd.DataFrame(columns = ['lifetime', 'mean'])

for column in df.columns:
    if column == 'voltage':
        continue
    else:
        try:
            
            
            mean_dQ = df.loc[mask, column].mean()

            cell_lifetime = get_lifetime.lifetime(str(column))

            df_correlation.iloc[len(df_correlation)] = [cell_lifetime, mean_dQ]

        except KeyError:
            print(f'{column} couldn\'t be accessed')

        
            





        


    

    
```
